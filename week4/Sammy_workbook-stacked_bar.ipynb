```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from prettytable import PrettyTable
```

```python
#Student Performance Data Set
students = pd.read_csv('StudentPerformanceFactors.csv', sep=",")
```

```python
students.head()
```

```python
import random

sleep = students["Sleep_Hours"]
score = students["Exam_Score"]

#filter for just one gender:
male = students[students['Gender'] == "Male"]
sleep = male["Sleep_Hours"]
score = male["Exam_Score"]

print("Number of Students", len(male))

plt.figure(figsize=(30,16))

mod_sleep = []

for s in sleep:
    mod = s + float(random.randint(-50, 50)/150)
    mod_sleep.append(mod)

# Use numpy to do a simple linear fit to time series data 
model = np.polyfit(score, mod_sleep, 1)
predicted = np.polyval(model, score)

plt.plot(score, predicted, linewidth=2, color='black')

slope = model[0]
intercept = model[1]

plt.scatter(score, mod_sleep, s=8, alpha=0.75)

plt.xlabel("Exam Score", fontsize=24)
plt.ylabel("Sleep Hours", fontsize=24)

plt.annotate('$y=%3.5sx+%3.5s$'%(slope, intercept), xy=(0.02, 0.95), xycoords='axes fraction', fontsize=14)

r2 = np.corrcoef(mod_sleep, score)[0, 1]**2
plt.annotate(r'$R^2=%.5f$' % r2, xy=(0.02, 0.9), xycoords='axes fraction', fontsize=14)
```

```python
import matplotlib.pyplot as plt
import numpy as np

parent = students["Parental_Involvement"]
score = students["Exam_Score"]

# Sample data
categories = ['A', 'B', 'C', 'D', 'F']

low = [0,0,0,0,0]
med = [0,0,0,0,0]
high = [0,0,0,0,0]

low_count = [0,0,0,0,0]
med_count = [0,0,0,0,0]
high_count = [0,0,0,0,0]

for i in range(0, len(score)):
    if score.iloc[i]>89 and score.iloc[i]<101:
        if parent.iloc[i] == "Low":
            low[0] = low[0] + score.iloc[i]
            low_count[0] = low_count[0] + 1
        elif parent.iloc[i] == "High":
            high[0] = high[0] + score.iloc[i]
            high_count[0] = high_count[0] + 1
        else:
            med[0] = med[0] + score.iloc[i]
            med_count[0] = med_count[0] + 1
    elif score.iloc[i]>79 and score.iloc[i]<90:
        if parent.iloc[i] == "Low":
            low[1] = low[1] + score.iloc[i]
            low_count[1] = low_count[1] + 1
        elif parent.iloc[i] == "High":
            high[1] = high[1] + score.iloc[i]
            high_count[1] = high_count[1] + 1
        else:
            med[1] = med[1] + score.iloc[i]
            med_count[1] = med_count[1] + 1
    elif score.iloc[i]>69 and score.iloc[i]<80:
        if parent.iloc[i] == "Low":
            low[2] = low[2] + score.iloc[i]
            low_count[2] = low_count[2] + 1
        elif parent.iloc[i] == "High":
            high[2] = high[2] + score.iloc[i]
            high_count[2] = high_count[2] + 1
        else:
            med[2] = med[2] + score.iloc[i]
            med_count[2] = med_count[2] + 1            
    elif score.iloc[i]>59 and score.iloc[i]<70:
        if parent.iloc[i] == "Low":
            low[3] = low[3] + score.iloc[i]
            low_count[3] = low_count[3] + 1
        elif parent.iloc[i] == "High":
            high[3] = high[3] + score.iloc[i]
            high_count[3] = high_count[3] + 1
        else:
            med[3] = med[3] + score.iloc[i]
            med_count[3] = med_count[3] + 1
    else:
        if parent.iloc[i] == "Low":
            low[4] = low[4] + score.iloc[i]
            low_count[4] = low_count[4] + 1
        elif parent.iloc[i] == "High":
            high[4] = high[4] + score.iloc[i]
            high_count[4] = high_count[4] + 1
        else:
            med[4] = med[4] + score.iloc[i]
            med_count[4] = med_count[4] + 1
            
mph_count = [0, 0, 0, 0, 0]

for l in range(0,5):
    mph_count[l] = low_count[l] + med_count[l]
            
print(len(low), len(med), len(high))
            
# Create the first layer (bottom)
plt.bar(categories, low_count, label='Low')

# Create the second layer on top of the first
# The 'bottom' argument is the height of the previous layer
plt.bar(categories, med_count, bottom=low_count, label='Medium')

# Create the third layer on top of the second
# The 'bottom' argument is the cumulative sum of the previous layers
plt.bar(categories, high_count, bottom=mph_count, label='High')

# Add labels, title, and legend
plt.xlabel("Exam Grade", fontsize=24)
plt.ylabel("Total Parental Involvement", fontsize=18)
plt.title('Stacked Bar of Exam Scores')
plt.legend()
plt.show()
```

```python
mph = [0, 0, 0, 0, 0]

for l in range(0,5):
    low[l] = low_count[l]/(low_count[l]+med_count[l]+high_count[l])
    med[l] = med_count[l]/(low_count[l]+med_count[l]+high_count[l])
    high[l] = high_count[l]/(low_count[l]+med_count[l]+high_count[l])
    mph[l] = low[l] + med[l]
    
    print(low[l], med[l], high[l])

# Create the first layer (bottom)
plt.bar(categories, low, label='Low')

# Create the second layer on top of the first
# The 'bottom' argument is the height of the previous layer
plt.bar(categories, med, bottom=low, label='Medium')

# Create the third layer on top of the second
# The 'bottom' argument is the cumulative sum of the previous layers
plt.bar(categories, high, bottom=mph, label='High')

# Add labels, title, and legend
plt.xlabel("Exam Grade", fontsize=24)
plt.ylabel("Total Parental Involvement", fontsize=18)
plt.title('Stacked Bar of Exam Score Percentages')
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Sample data
categories = ['A', 'B', 'C', 'D']
data1 = np.array([10, 15, 7, 12])
data2 = np.array([5, 8, 10, 6])
data3 = np.array([3, 4, 6, 8])

# Create the first layer (bottom)
plt.bar(categories, data1, label='Category 1')

# Create the second layer on top of the first
# The 'bottom' argument is the height of the previous layer
plt.bar(categories, data2, bottom=data1, label='Category 2')

# Create the third layer on top of the second
# The 'bottom' argument is the cumulative sum of the previous layers
plt.bar(categories, data3, bottom=data1 + data2, label='Category 3')

# Add labels, title, and legend
plt.xlabel('Groups')
plt.ylabel('Values')
plt.title('Stacked Bar Chart Example')
plt.legend()
plt.show()
```
