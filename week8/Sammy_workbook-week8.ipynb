Can we predict Social Media Hours?

1. Join with data set of student habits to get social media use
2. Create new model with social media hours predicted

```python
'''
Create an array of social media hours for all students, 
based on relationship of social media hour behavior correlated to exam score
'''

import random

social_media_minutes = []
j = 0

for e in range (0, len(exam)):
    done = False
    while done == False:
        i = random.randint(0, len(habits.exam_score)-1)
        if (exam[j] >= 100):
            social_media_minutes.append(8)
            done = True
            j = j + 1
        elif (exam[j] == int(habits.exam_score[i])):
            social_media_minutes.append(int(habits.social_media_hours[i]*60))
            done = True
            j = j + 1
```

```python
#Look at the statistical correlation to exam score

score = []
minutes = []

#filter for just one gender (note: no significant difference between genders):
for i in range(0, len(students['Gender'])):
    if (students['Gender'][i] == 1):
        score.append(students["Exam_Score"][i])
        minutes.append(social_media_minutes[i])

plt.figure(figsize=(30,16))

plt.scatter(score, minutes, s=8, alpha=0.75)

plt.xlabel("Exam Score", fontsize=24)
plt.ylabel("Social Media", fontsize=24)

# Use numpy to do a simple linear fit to time series data 
model = np.polyfit(score, minutes, 1)
predicted = np.polyval(model, score)

plt.plot(score, predicted, linewidth=2, color='black')

slope = model[0]
intercept = model[1]
plt.annotate('$y=%3.5sx+%3.5s$'%(slope, intercept), xy=(0.02, 0.95), xycoords='axes fraction', fontsize=14)

r2 = np.corrcoef(minutes, score)[0, 1]**2
plt.annotate(r'$R^2=%.5f$' % r2, xy=(0.02, 0.9), xycoords='axes fraction', fontsize=14)
```

```python
#verify that social_media_minutes is a valid 1-D array of integer values

print(social_media_minutes)
```

```python
from sklearn.model_selection import train_test_split

# Split dataset into training set and test set
X_train, X_test, y_train, y_test = train_test_split(students, social_media_minutes, test_size=0.3) # 70% training and 30% test

n_samples, n_features = X_train.shape

#print training set shape
print(X_train.shape)

#print training set shape
print(X_test.shape)
```
