```python
import tkinter as tk
from tkinter import ttk
from tkinter import messagebox
from sklearn.metrics import classification_report
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
import pandas as pd
from sklearn.preprocessing import LabelEncoder
import numpy as np

value_mapping = {
    "Low": 0, "Medium": 1, "High": 2,
    "Yes": 1, "No": 0,
    "Public": 0, "Private": 1,
    "Negative": 0, "Neutral": 1, "Positive": 2,
    "High School": 0, "College": 1, "Postgraduate": 2,
    "Near": 0, "Moderate": 1, "Far": 2,
    "Male": 0, "Female": 1
}

# Load dataset
df = pd.read_csv("StudentPerformanceFactors.csv")

# Replace values
df = df.replace(value_mapping)

# Extract features by dropping the target column 'Exam_Score'
X = df.drop(columns=['Exam_Score'])

# Extract target labels
y = df['Exam_Score']

# Save new CSV
df.to_csv("StudentPerformanceFactors_dataset_numeric.csv", index=False)

# read dataset
students=pd.read_csv("StudentPerformanceFactors_dataset_numeric.csv")

exam_score=pd.read_csv("ExamScore.csv")
scoreresult = []

for i in range(0, len(exam_score.Exam_Score)):
    scoreresult.append(exam_score.Exam_Score[i])

# Split dataset into training set and test set
X_train, X_test, y_train, y_test = train_test_split(students, scoreresult, test_size=0.3) # 70% training and 30% test
clf = RandomForestClassifier(n_estimators=100, random_state=42)
clf.fit(X_train, y_train)
n_samples, n_features = X_train.shape

# Scroll function
def _on_mousewheel(event):
    canvas.yview_scroll(int(-1 * (event.delta / 120)), "units")

# Main window
window = tk.Tk()
window.title('Student Performance Evaluator')
window.geometry('800x500')

# Main frame for two columns
main_frame = ttk.Frame(window)
main_frame.pack(fill='both', expand=1)

# Left scrollable form frame
form_container = ttk.Frame(main_frame)
form_container.pack(side='left', fill='both', expand=1)

canvas = tk.Canvas(form_container)
canvas.pack(side='left', fill='both', expand=1)

scrollbar = ttk.Scrollbar(form_container, orient='vertical', command=canvas.yview)
scrollbar.pack(side='right', fill='y')

canvas.configure(yscrollcommand=scrollbar.set)
canvas.bind('<Configure>', lambda e: canvas.configure(scrollregion=canvas.bbox("all")))

frame2 = ttk.Frame(canvas)
canvas.create_window((0, 0), window=frame2, anchor='nw')

canvas.bind_all("<MouseWheel>", _on_mousewheel)

# Right messages panel
messages_frame = ttk.Frame(main_frame, width=220)
messages_frame.pack(side='right', fill='y')
messages_frame.pack_propagate(False)

ttk.Label(messages_frame, text="Messages", font=('Arial', 12, 'bold')).pack(anchor='w', padx=10, pady=5)

def add_message(sender, text, color="black"):
    label = ttk.Label(messages_frame, text=f"{sender}: {text}", foreground=color, wraplength=200)
    label.pack(anchor='w', pady=2)

fields = []

# Title
title_label = ttk.Label(master=frame2, text='Student Performance Evaluator', font='Roboto 24 bold')
title_label.pack()

# Helper for entry
def add_entry(label_text):
    ttk.Label(frame2, text=label_text).pack(anchor='w', padx=20, pady=(10, 0))
    entry = ttk.Entry(frame2)
    entry.pack(anchor='w', padx=20, pady=5)
    fields.append(entry)

# Helper for dropdown
def add_dropdown(label_text, variable, *options):
    ttk.Label(frame2, text=label_text).pack(anchor='w', padx=20, pady=(10, 0))
    ttk.OptionMenu(frame2, variable, options[0], *options).pack(anchor='w', padx=20, pady=5)
    fields.append(variable)

# Entries
add_entry('Hours_Studied')
add_entry('Attendance')

# Dropdowns
parental_involvement_var = tk.StringVar(value="Medium")
add_dropdown('Parental_Involvement', parental_involvement_var, "Low", "Medium", "High")

access_to_resources_var = tk.StringVar(value="Medium")
add_dropdown('Access_to_Resources', access_to_resources_var, "Low", "Medium", "High")

extracurricular_activities_var = tk.StringVar(value="Yes")
add_dropdown('Extracurricular_Activities', extracurricular_activities_var, "Yes", "No")

add_entry('Sleep_Hours')
add_entry('Previous_Scores')

motivation_level_var = tk.StringVar(value="Medium")
add_dropdown('Motivation_Level', motivation_level_var, "Low", "Medium", "High")

internet_access_var = tk.StringVar(value="Yes")
add_dropdown('Internet_Access', internet_access_var, "Yes", "No")

add_entry('Tutoring_Sessions')

family_income_var = tk.StringVar(value="Medium")
add_dropdown('Family_Income', family_income_var, "Low", "Medium", "High")

teacher_quality_var = tk.StringVar(value="Medium")
add_dropdown('Teacher_Quality', teacher_quality_var, "Low", "Medium", "High")

school_type_var = tk.StringVar(value="Public")
add_dropdown('School_Type', school_type_var, "Public", "Private")

peer_influence_var = tk.StringVar(value="Neutral")
add_dropdown('Peer_Influence', peer_influence_var, "Negative", "Neutral", "Positive")

add_entry('Physical_Activity')

learning_disabilities_var = tk.StringVar(value="No")
add_dropdown('Learning_Disabilities', learning_disabilities_var, "No", "Yes")

parental_education_level_var = tk.StringVar(value="High School")
add_dropdown('Parental_Education_Level', parental_education_level_var, "High School", "College", "Postgraduate")

distance_from_home_var = tk.StringVar(value="Moderate")
add_dropdown('Distance_from_Home', distance_from_home_var, "Near", "Moderate", "Far")

gender_var = tk.StringVar(value="Male")
add_dropdown('Gender', gender_var, "Male", "Female")

# Save function
data = []
def save():
    data.clear()
    for field in fields:
        if isinstance(field, ttk.Entry):
            val = field.get()
            if val.strip() == "":
                add_message("System", "Error: Please fill in all fields.", "red")
                return
            data.append(float(val))
        elif isinstance(field, tk.StringVar):
            val = field.get()
            data.append(value_mapping.get(val, -1))

    with open("case.txt", "a") as file:
        file.write(str(data) + "\n")

    add_message("System", "Data saved successfully.", "green")



# predict fucntions
def predict():
    inputs = []
    for field in fields:
        if isinstance(field, tk.Entry) or isinstance(field, ttk.Entry):
            val = field.get()
            if val.strip() == "":
                messagebox.showerror("Error", "Please fill in all fields before predicting.")
                return
            try:
                inputs.append(float(val))
            except ValueError:
                messagebox.showerror("Error", f"Invalid number: '{val}'")
                return
        elif isinstance(field, tk.StringVar):
            val = field.get()
            if val not in value_mapping:
                messagebox.showerror("Error", f"Invalid dropdown value: {val}")
                return
            inputs.append(value_mapping[val])

    # Convert to numpy array for prediction
    inputs_array = np.array(inputs).reshape(1, -1)
    prediction = clf.predict(inputs_array)[0]
    messagebox.showinfo("Prediction Result", f"Predicted Exam Grade: {prediction}")

# Buttons
predict_button = ttk.Button(master=frame2, text='Predict',command=predict)
predict_button.pack(anchor='w', padx=20, pady=5)

save_button = ttk.Button(frame2, text="Save", command=save)
save_button.pack(padx=20, pady=5, anchor='w')

# Run
window.mainloop()
```
