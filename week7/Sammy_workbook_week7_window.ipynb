```python
import tkinter as tk
from tkinter import ttk
from tkinter import messagebox

def _on_mousewheel(event):
    canvas.yview_scroll(int(-1 * (event.delta / 120)), "units")

# window 
window = tk.Tk()
window.title('Student Performance Evaluator')
window.geometry('600x400')

# create main frame
main_frame = ttk.Frame(master = window)
main_frame.pack(fill='both', expand = 1)

# create canvas
canvas = tk.Canvas(master = main_frame)
canvas.pack(side = 'left', fill='both', expand = 1)

# scrollbar
scrollbar = ttk.Scrollbar(master = main_frame, orient = 'vertical', command = canvas.yview)
scrollbar.pack(side = 'right', fill = 'y')

# configure the canvas
canvas.configure(yscrollcommand = scrollbar.set)
canvas.bind('<Configure>', lambda e: canvas.configure(scrollregion = canvas.bbox("all")))                          

# add another frame inside canvas
frame2 = ttk.Frame(master = canvas)
# add frame to window in canvas
canvas.create_window((0,0), window = frame2, anchor = 'nw')

# bind mousewheel to canvas (Windows/Mac)
canvas.bind_all("<MouseWheel>", _on_mousewheel)

# title
title_label = ttk.Label(master = frame2, text = 'Student Performance Evaluator', font = 'Roboto 24 bold')
title_label.pack()

# input 
input_frame1 = ttk.Frame(master = frame2)
input_frame1.pack()

# Dictionary to map dropdown values to numbers
value_mapping = {
    "Low": 0,
    "Medium": 1,
    "High": 2,
    "Yes": 1,
    "No": 0,
    "Public": 0,
    "Private": 1,
    "Negative": 0,
    "Neutral": 1,
    "Positive": 2,
    "High School": 0,
    "College": 1,
    "Postgraduate": 2,
    "Near": 0,
    "Moderate": 1,
    "Far": 2,
    "Male": 0,
    "Female": 1
}

# Entry and OptionMenu fields
fields = []

# Entry for Hours_Studied
fields.append(ttk.Entry(frame2))
ttk.Label(frame2, text='Hours_Studied').pack(anchor='w', padx=20, pady=(10, 0))
fields[-1].pack(anchor='w', padx=20, pady=5)

# Entry for Attendance
fields.append(ttk.Entry(frame2))
ttk.Label(frame2, text='Attendance').pack(anchor='w', padx=20, pady=(10, 0))
fields[-1].pack(anchor='w', padx=20, pady=5)

# Dropdown for Parental_Involvement
parental_involvement_var = tk.StringVar(value="Medium")
ttk.Label(frame2, text='Parental_Involvement').pack(anchor='w', padx=20, pady=(10, 0))
ttk.OptionMenu(frame2, parental_involvement_var, "Medium", "Low", "Medium", "High").pack(anchor='w', padx=20, pady=5)
fields.append(parental_involvement_var)

# Dropdown for Access_to_Resources
access_to_resources_var = tk.StringVar(value="Medium")
ttk.Label(frame2, text='Access_to_Resources').pack(anchor='w', padx=20, pady=(10, 0))
ttk.OptionMenu(frame2, access_to_resources_var, "Medium", "Low", "Medium", "High").pack(anchor='w', padx=20, pady=5)
fields.append(access_to_resources_var)

# Dropdown for Extracurricular_Activities
extracurricular_activities_var = tk.StringVar(value="Yes")
ttk.Label(frame2, text='Extracurricular_Activities').pack(anchor='w', padx=20, pady=(10, 0))
ttk.OptionMenu(frame2, extracurricular_activities_var, "Yes", "No", "Yes").pack(anchor='w', padx=20, pady=5)
fields.append(extracurricular_activities_var)

# Entry for Sleep_Hours
fields.append(ttk.Entry(frame2))
ttk.Label(frame2, text='Sleep_Hours').pack(anchor='w', padx=20, pady=(10, 0))
fields[-1].pack(anchor='w', padx=20, pady=5)

# Entry for Previous_Scores
fields.append(ttk.Entry(frame2))
ttk.Label(frame2, text='Previous_Scores').pack(anchor='w', padx=20, pady=(10, 0))
fields[-1].pack(anchor='w', padx=20, pady=5)

# Dropdown for Motivation_Level
motivation_level_var = tk.StringVar(value="Medium")
ttk.Label(frame2, text='Motivation_Level').pack(anchor='w', padx=20, pady=(10, 0))
ttk.OptionMenu(frame2, motivation_level_var, "Medium", "Low", "Medium", "High").pack(anchor='w', padx=20, pady=5)
fields.append(motivation_level_var)

# Dropdown for Internet_Access
internet_access_var = tk.StringVar(value="Yes")
ttk.Label(frame2, text='Internet_Access').pack(anchor='w', padx=20, pady=(10, 0))
ttk.OptionMenu(frame2, internet_access_var, "Yes", "No", "Yes").pack(anchor='w', padx=20, pady=5)
fields.append(internet_access_var)

# Entry for Tutoring_Sessions
fields.append(ttk.Entry(frame2))
ttk.Label(frame2, text='Tutoring_Sessions').pack(anchor='w', padx=20, pady=(10, 0))
fields[-1].pack(anchor='w', padx=20, pady=5)

# Dropdown for Family_Income
family_income_var = tk.StringVar(value="Medium")
ttk.Label(frame2, text='Family_Income').pack(anchor='w', padx=20, pady=(10, 0))
ttk.OptionMenu(frame2, family_income_var, "Medium", "Low", "Medium", "High").pack(anchor='w', padx=20, pady=5)
fields.append(family_income_var)

# Dropdown for Teacher_Quality
teacher_quality_var = tk.StringVar(value="Medium")
ttk.Label(frame2, text='Teacher_Quality').pack(anchor='w', padx=20, pady=(10, 0))
ttk.OptionMenu(frame2, teacher_quality_var, "Medium", "Low", "Medium", "High").pack(anchor='w', padx=20, pady=5)
fields.append(teacher_quality_var)

# Dropdown for School_Type
school_type_var = tk.StringVar(value="Public")
ttk.Label(frame2, text='School_Type').pack(anchor='w', padx=20, pady=(10, 0))
ttk.OptionMenu(frame2, school_type_var, "Public", "Private", "Public").pack(anchor='w', padx=20, pady=5)
fields.append(school_type_var)

# Dropdown for Peer_Influence
peer_influence_var = tk.StringVar(value="Neutral")
ttk.Label(frame2, text='Peer_Influence').pack(anchor='w', padx=20, pady=(10, 0))
ttk.OptionMenu(frame2, peer_influence_var, "Neutral", "Negative", "Neutral", "Positive").pack(anchor='w', padx=20, pady=5)
fields.append(peer_influence_var)

# Entry for Physical_Activity
fields.append(ttk.Entry(frame2))
ttk.Label(frame2, text='Physical_Activity').pack(anchor='w', padx=20, pady=(10, 0))
fields[-1].pack(anchor='w', padx=20, pady=5)

# Dropdown for Learning_Disabilities
learning_disabilities_var = tk.StringVar(value="No")
ttk.Label(frame2, text='Learning_Disabilities').pack(anchor='w', padx=20, pady=(10, 0))
ttk.OptionMenu(frame2, learning_disabilities_var, "No", "Yes", "No").pack(anchor='w', padx=20, pady=5)
fields.append(learning_disabilities_var)

# Dropdown for Parental_Education_Level
parental_education_level_var = tk.StringVar(value="High School")
ttk.Label(frame2, text='Parental_Education_Level').pack(anchor='w', padx=20, pady=(10, 0))
ttk.OptionMenu(frame2, parental_education_level_var, "High School", "High School", "College", "Postgraduate").pack(anchor='w', padx=20, pady=5)
fields.append(parental_education_level_var)

# Dropdown for Distance_from_Home
distance_from_home_var = tk.StringVar(value="Moderate")
ttk.Label(frame2, text='Distance_from_Home').pack(anchor='w', padx=20, pady=(10, 0))
ttk.OptionMenu(frame2, distance_from_home_var, "Moderate", "Near", "Moderate", "Far").pack(anchor='w', padx=20, pady=5)
fields.append(distance_from_home_var)

# Dropdown for Gender
gender_var = tk.StringVar(value="Male")
ttk.Label(frame2, text='Gender').pack(anchor='w', padx=20, pady=(10, 0))
ttk.OptionMenu(frame2, gender_var, "Male", "Male", "Female").pack(anchor='w', padx=20, pady=5)
fields.append(gender_var)

# Save function
def save():
    for field in fields:
        if isinstance(field, ttk.Entry):
            val = field.get()
            if val.strip() == "":
                messagebox.showerror("Error", "Please fill in all fields.")
                return
            data.append(float(val))
        elif isinstance(field, tk.StringVar):
            val = field.get()
            data.append(value_mapping.get(val, -1))

    with open("case.txt", "a") as file:
        file.write(str(data) + "\n")
    messagebox.showinfo("Saved", "Data saved successfully.")

predict_button = ttk.Button(master = frame2, text = 'Predict') 
predict_button.pack(anchor='w', padx=20, pady=5)

save_button = ttk.Button(frame2, text="Save", command=save).pack(pady=10, anchor='w')

# run
window.mainloop()
```
