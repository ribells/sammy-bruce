```python
import tkinter as tk
from tkinter import ttk

def _on_mousewheel(event):
    canvas.yview_scroll(int(-1 * (event.delta / 120)), "units")

# window 
window = tk.Tk()
window.title('Student Performance Evaluator')
window.geometry('600x400')

# create main frame
main_frame = ttk.Frame(master = window)
main_frame.pack(fill='both', expand = 1)

# create canvas
canvas = tk.Canvas(master = main_frame)
canvas.pack(side = 'left', fill='both', expand = 1)

# scrollbar
scrollbar = ttk.Scrollbar(master = main_frame, orient = 'vertical', command = canvas.yview)
scrollbar.pack(side = 'right', fill = 'y')

# configure the canvas
canvas.configure(yscrollcommand = scrollbar.set)
canvas.bind('<Configure>', lambda e: canvas.configure(scrollregion = canvas.bbox("all")))                          

# add another frame inside canvas
frame2 = ttk.Frame(master = canvas)
# add frame to window in canvas
canvas.create_window((0,0), window = frame2, anchor = 'nw')

# bind mousewheel to canvas (Windows/Mac)
canvas.bind_all("<MouseWheel>", _on_mousewheel)

# title
title_label = ttk.Label(master = frame2, text = 'Student Performance Evaluator', font = 'Roboto 24 bold')
title_label.pack()

# input 
input_frame1 = ttk.Frame(master = frame2)
input_frame1.pack()

# entry 1
ttk.Label(frame2, text="Entry 1:").pack(anchor='w', padx=20, pady=(10, 0))
entry_frame1 = ttk.Entry(master = frame2)
entry_frame1.pack(anchor='w', padx=20, pady=5)

# entry 2
ttk.Label(frame2, text="Entry 2:").pack(anchor='w', padx=20, pady=(0, 0))
entry_frame2 = ttk.Entry(master = frame2)
entry_frame2.pack(anchor='w', padx=20, pady=5)

# entry 3
ttk.Label(frame2, text="Entry 3:").pack(anchor='w', padx=20, pady=(0, 0))
entry_frame3 = ttk.Entry(master = frame2)
entry_frame3.pack(anchor='w', padx=20, pady=5)


predict_button = ttk.Button(master = frame2, text = 'Predict') 
predict_button.pack(anchor='w', padx=20, pady=5)

# run
window.mainloop()
```
