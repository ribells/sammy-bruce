```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from prettytable import PrettyTable
```

```python
#Student Performance Data Set
students = pd.read_csv('student_habits_performance.csv', sep=",")
```

```python
students.head()
```

```python
import random

sleep = students["Sleep_Hours"]
score = students["Exam_Score"]

#filter for just one gender:
male = students[students['Gender'] == "Male"]
sleep = male["Sleep_Hours"]
score = male["Exam_Score"]

print("Number of Students", len(male))

plt.figure(figsize=(30,16))

mod_sleep = []

for s in sleep:
    mod = s + float(random.randint(-50, 50)/150)
    mod_sleep.append(mod)

# Use numpy to do a simple linear fit to time series data 
model = np.polyfit(score, mod_sleep, 1)
predicted = np.polyval(model, score)

plt.plot(score, predicted, linewidth=2, color='black')

slope = model[0]
intercept = model[1]

plt.scatter(score, mod_sleep, s=8, alpha=0.75)

plt.xlabel("Exam Score", fontsize=24)
plt.ylabel("Sleep Hours", fontsize=24)

plt.annotate('$y=%3.5sx+%3.5s$'%(slope, intercept), xy=(0.02, 0.95), xycoords='axes fraction', fontsize=14)

r2 = np.corrcoef(mod_sleep, score)[0, 1]**2
plt.annotate(r'$R^2=%.5f$' % r2, xy=(0.02, 0.9), xycoords='axes fraction', fontsize=14)
```

```python
# Convert diet_quality categories to numeric values
def convert_diet(value):
    if value == "Poor":
        return 0
    elif value == "Fair":
        return 1
    elif value == "Good":
        return 2
    else:
        return np.nan  #For other values

students["diet_quality_numeric"] = students["diet_quality"].apply(convert_diet)

diet = students["diet_quality_numeric"]
score = students["exam_score"]

print("Number of Students", len(students))

plt.figure(figsize=(30,16))

plt.scatter(score, diet, s=8, alpha=0.75)

plt.xlabel("exam score", fontsize=24)
plt.ylabel("diet quality", fontsize=24)

# Drop NaNs before fitting to avoid errors
valid = (~diet.isna()) & (~score.isna())
diet_clean = diet[valid]
score_clean = score[valid]

model = np.polyfit(score_clean, diet_clean, 1)
predicted = np.polyval(model, score_clean)

plt.plot(score_clean, predicted, linewidth=2, color='black')

slope = model[0]
intercept = model[1]
plt.annotate('$y=%3.5sx+%3.5s$' % (slope, intercept), xy=(0.02, 0.95), xycoords='axes fraction', fontsize=14)

r2 = np.corrcoef(diet_clean, score_clean)[0, 1]**2
plt.annotate(r'$R^2=%.5f$' % r2, xy=(0.02, 0.9), xycoords='axes fraction', fontsize=14)
```
